## 5. Data Cleaning and Preprocessing

Prepare the dataset for analysis.


```python
df.isnull().sum()
```

```python
cols_to_fill = ["director", "cast", "country", "rating"]

df[cols_to_fill] = df[cols_to_fill].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"].str.strip(), errors="coerce")
```

```python
df.head()
```

```python
df["rating"].unique()
```

```python
df["rating"].value_counts()
```

```python
mask = df["rating"].str.contains(
    r"min|season",
    case=False,
    na=False
)
df.loc[mask, "duration"]=df.loc[mask, "rating"]
df.loc[mask, "rating"]=np.nan 
```

```python
df["rating"].value_counts()
```

```python
df["duration"].isna().sum()
```

```python
df[df["rating"].isna()]
```

```python
df["rating"]=df["rating"].fillna("Unknown")
```

```python
df["rating"].isna().sum()
```


### Cleaning Summary

- Missing values were handled individually based on the nature and importance of each column.
- Missing values in `director`, `country`, and `cast` were replaced with `"Unknown"` to preserve records without introducing fabricated information.
- The `date_added` column was converted from `object` to `datetime` to support date-based analysis.
- No duplicate records were found.
- Identified incorrect values in the `rating` column, where some duration values had been mistakenly stored.
- Recovered the misplaced duration information by transferring it from `rating` to `duration`.
- The resulting missing rating values were replaced with `"Unknown"`.
- Missing values in `date_added` were retained because their original dates could not be reliably determined.
- The three missing values in `duration` were investigated and addressed by recovering the misplaced information from `rating`.


## 6. Exploratory Data Analysis (EDA)

**Research Question:** Which type of content is more common on Netflix?

```python

colors = ["#E50914", "#8B0000"]
counts = df["type"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    counts,
    labels=counts.index,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"edgecolor": "white", "linewidth": 1.5},
)
plt.title("Distribution by Type")
plt.show()
```

**Research Question:** What are the most frequently listed genres on Netflix?

```python
top_genres = (
    df["listed_in"].str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_genres.values, y=top_genres.index, color="#E50914")
plt.title("Top 10 Genres on Netflix")
plt.xlabel("Number of titles")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
```

**Research Question:** Which countries contribute the most titles to Netflix?

```python
country = (
    df["country"].str.split(",").explode().str.strip()
    .loc[lambda s: s != "Unknown"]
    .value_counts().head(10)
)
plt.figure(figsize=(10, 6))
sns.barplot(x=country.values, y=country.index, color="#E50914")
plt.title("Top 10 Countries on Netflix")
plt.xlabel("Number of titles")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

**Research Question:** How has Netflix's content library evolved over the years?

```python
df["year_added"] = df["date_added"].dt.year

yearly = (
    df.dropna(subset=["year_added"])
    .groupby(["year_added", "type"])
    .size()
    .unstack(fill_value=0)
)

plt.figure(figsize=(10, 6))
plt.plot(yearly.index, yearly["Movie"], color="#E50914", label="Movies")
plt.plot(yearly.index, yearly["TV Show"], color="#F28B91", label="TV Shows")
plt.title("Titles Added to Netflix per Year")
plt.xlabel("Year added")
plt.ylabel("Number of titles")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

**Research Question:** What are the most common content ratings?

```python
df["rating"].value_counts().head()
```

```python
order = df["rating"].value_counts().head().index

plt.figure(figsize=(8, 5))
sns.countplot(data=df, y="rating", order=order, hue="rating",
              hue_order=order, palette="Reds_r", legend=False)
plt.title("Count by Rating")
plt.xlabel("Count")
plt.ylabel("Rating")
plt.tight_layout()
plt.show()
```

**Research Questions:**
- What is the distribution of movie durations?
- How many seasons do TV Shows typically have?

```python
movies = df[df["type"] == "Movie"].copy()
movies["minutes"] = movies["duration"].str.extract(r"(\d+)").astype(int)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(movies["minutes"], bins=30, color="#E50914", ax=axes[0])
axes[0].set_title("Distribution of Movie Durations")
axes[0].set_xlabel("Duration (minutes)")
axes[0].set_ylabel("Number of movies")

sns.boxplot(x=movies["minutes"], color="#F28B91", ax=axes[1])
axes[1].set_title("Movie Duration: Median and Outliers")
axes[1].set_xlabel("Duration (minutes)")

plt.tight_layout()
plt.show()

print(movies["minutes"].describe())
```

```python
df["duration"].nunique()
```

```python
shows = df[df["type"] == "TV Show"].copy()

# "duration" is text like "2 Seasons". Pull out just the number (\d+ = one or more digits) and convert it to an integer so we can sort and plot it
shows["seasons"] = shows["duration"].str.extract(r"(\d+)").astype(int) 

seasons = shows["seasons"].value_counts().sort_index()

plt.figure(figsize=(10, 6))
sns.barplot(x=seasons.index, y=seasons.values, color="#E50914")
plt.title("Number of Seasons per TV Show")
plt.xlabel("Number of seasons")
plt.ylabel("Number of TV shows")
plt.tight_layout()
plt.show()

print((seasons / seasons.sum() * 100).round(1))
```

## 7. Key Insights and Conclusions

1. **Movies dominate the catalog.** Movies make up 69.6% of titles (6,131) and TV Shows 30.4% (2,676).
2. **Most TV shows are short-lived.** 67.0% of TV shows have only one season, and 90.3% have three or fewer. The longest has 17 seasons.

4. **A few countries supply most content.** The United States leads with 2,818 titles, followed by India (972) and the United Kingdom (419).
5. **The library is mostly recent.** Half of all titles were released in 2017 or later, and 75% in 2013 or later, although the oldest dates back to 1925.

### Final Conclusion

Netflix's catalog is built mostly from recent Movies, with TV Shows making up about a third of titles and most of those running a single season. The content is aimed mainly at teen and adult viewers, and a small group of countries, led by the United States and India, supplies a large share of titles. ## 8. Limitations

- **Missing data:** `director` is missing for 29.9% of titles, and `country` and `cast` for about 9.4% each. These were filled with "Unknown", so rankings by director or country exclude them.
- **Data entry errors:** three rows had durations stored in the `rating` column. I fixed them, but other undetected errors may exist.
- **Time coverage:** the dataset ends in 2021, so 2021 is a partial year and the data may not reflect Netflix's current catalog.
- **Multi-valued columns:** titles can list several countries or genres, so each title is counted once per country or genre, and the totals exceed the number of titles.
- **No viewing data:** the dataset has no viewership, popularity or ratings data, so the results describe catalog composition, not what viewers watch.
